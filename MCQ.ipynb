```python
#Q1

import pandas as pd

path = "/Users/huynhsarah0712/Downloads/datathon-2026-round-1"
orders = pd.read_csv(f"{path}/orders.csv")
orders['order_date'] = pd.to_datetime(orders['order_date'])
orders = orders.sort_values(['customer_id', 'order_date'])
orders['gap_days'] = orders.groupby('customer_id')['order_date'].diff().dt.days
gaps = orders['gap_days'].dropna()
median_gap = gaps.median()
print("Median inter-order gap:", median_gap)
```

```python
#Q2

products = pd.read_csv(f"{path}/products.csv")
products['margin'] = (products['price'] - products['cogs']) / products['price']
segment_margin = products.groupby('segment')['margin'].mean()
print(segment_margin.sort_values(ascending=False))
best_segment = segment_margin.idxmax()
print(best_segment)
```

```python
#Q3

returns = pd.read_csv(f"{path}/returns.csv")
products = pd.read_csv(f"{path}/products.csv")
df = returns.merge(products, on='product_id', how='left')
df_streetwear = df[df['category'] == 'Streetwear']
reason_counts = df_streetwear['return_reason'].value_counts()
print(reason_counts)
top_reason = reason_counts.idxmax()
print(top_reason)
```

```python
#Q4

web = pd.read_csv(f"{path}/web_traffic.csv")
source_bounce = web.groupby('traffic_source')['bounce_rate'].mean()
print(source_bounce.sort_values())
best_source = source_bounce.idxmin()
print(best_source)
```

```python
#Q5

order_items = pd.read_csv(f"{path}/order_items.csv")
total_rows = len(order_items)
promo_rows = order_items['promo_id'].notna().sum()
percentage = promo_rows / total_rows * 100
print(percentage)
```

```python
#Q6

customers = pd.read_csv(f"{path}/customers.csv")
orders = pd.read_csv(f"{path}/orders.csv")
customers = customers[customers['age_group'].notna()]
df = orders.merge(customers[['customer_id', 'age_group']], on='customer_id', how='inner')
orders_per_customer = df.groupby(['customer_id', 'age_group']).size().reset_index(name='order_count')
avg_orders = orders_per_customer.groupby('age_group')['order_count'].mean()
print(avg_orders.sort_values(ascending=False))
best_group = avg_orders.idxmax()
print(best_group)
```

```python
#Q7

sales = pd.read_csv(f"{path}/sales.csv")
orders = pd.read_csv(f"{path}/orders.csv")
geo = pd.read_csv(f"{path}/geography.csv")
sales['Date'] = pd.to_datetime(sales['Date'])
orders['order_date'] = pd.to_datetime(orders['order_date'])
df = orders.merge(sales, left_on='order_date', right_on='Date', how='inner')
df = df.merge(geo[['zip', 'region']], on='zip', how='left')
region_revenue = df.groupby('region')['Revenue'].sum()
print(region_revenue.sort_values(ascending=False))
best_region = region_revenue.idxmax()
print(best_region)
```

```python
#Q8

orders = pd.read_csv(f"{path}/orders.csv")
cancelled = orders[orders['order_status'] == 'cancelled']
payment_counts = cancelled['payment_method'].value_counts()
print(payment_counts)
top_payment = payment_counts.idxmax()
print(top_payment)
```

```python
#Q9

returns = pd.read_csv(f"{path}/returns.csv")
order_items = pd.read_csv(f"{path}/order_items.csv")
products = pd.read_csv(f"{path}/products.csv")
returns_df = returns.merge(products[['product_id', 'size']], on='product_id')
orders_df = order_items.merge(products[['product_id', 'size']], on='product_id')
returns_count = returns_df.groupby('size').size()
orders_count = orders_df.groupby('size').size()
return_rate = (returns_count / orders_count).sort_values(ascending=False)
print(return_rate)
best_size = return_rate.idxmax()
print(best_size)
```

```python
#Q10

payments = pd.read_csv(f"{path}/payments.csv")
avg_payment = payments.groupby('installments')['payment_value'].mean()
print(avg_payment.sort_values(ascending=False))
best_plan = avg_payment.idxmax()
print(best_plan)
```
